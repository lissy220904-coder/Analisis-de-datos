```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Generar datos sintéticos realistas de satélites LEO (Perigeo X y Apogeo Y)
np.random.seed(42)
N = 1200

# Perigeo centrado ~500 km y Apogeo correlacionado ligeramente mayor
X = np.random.normal(loc=550, scale=180, size=N)
X = np.clip(X, 200, 1800)  # Limitar rango LEO realista
Y = X + np.random.exponential(scale=60, size=N) + 10

print(f"✅ Datos cargados correctamente. Muestra: N = {len(X)} satélites LEO.")
```

```python
# ---------------------------------------------------------
# PUNTO 2: HISTOGRAMA DE UNA VARIABLE ALEATORIA (Perigeo X)
# ---------------------------------------------------------

# Calcular la media de la variable X
mu_x = np.mean(X)

plt.figure(figsize=(8, 4.5))

# Graficar histograma
plt.hist(X, bins=30, color='#1f77b4', edgecolor='black', alpha=0.8)

# Agregar la línea de la media
plt.axvline(mu_x, color='red', linestyle='--', linewidth=2, label=rf'Media $\mu_x = {mu_x:.2f}$ km')

# Personalizar gráfico
plt.title('Punto 2: Histograma del Perigeo ($X$)', fontsize=12, fontweight='bold')
plt.xlabel('Perigeo (km)')
plt.ylabel('Frecuencia (Número de satélites)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Mostrar la gráfica
plt.show()
```

```python
mu_y = np.mean(Y)
counts_y, bin_edges_y = np.histogram(Y, bins=25)
bin_centers_y = (bin_edges_y[:-1] + bin_edges_y[1:]) / 2
errors_y = np.sqrt(counts_y)

plt.figure(figsize=(7, 4))
plt.bar(
    bin_centers_y,
    counts_y,
    width=(bin_edges_y[1] - bin_edges_y[0]),
    yerr=errors_y,
    capsize=3,
    color="#17becf",  # Tono turquesa/cian para el apogeo
    edgecolor="black",
    alpha=0.75,
)
plt.axvline(
    mu_y, color="red", linestyle="--", linewidth=2, label=f"Media μ_y = {mu_y:.2f}"
)
plt.title("Histograma del Apogeo (Y) con Promedio y Errores")
plt.xlabel("Apogeo (km)")
plt.ylabel("Frecuencia")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
# ---------------------------------------------------------
# PUNTO 3: GRÁFICO DE DISPERSIÓN DE DOS VARIABLES (Perigeo X vs Apogeo Y)
# ---------------------------------------------------------

plt.figure(figsize=(7, 5))

# Graficar dispersión (scatter plot)
plt.scatter(X, Y, alpha=0.4, color='#1f77b4', edgecolors='none', s=20, label='Satélites LEO')

# Personalización
plt.title('Punto 3: Diagrama de Dispersión (Perigeo vs Apogeo)', fontsize=12, fontweight='bold')
plt.xlabel('Perigeo $X$ (km)')
plt.ylabel('Apogeo $Y$ (km)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Mostrar la gráfica
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# ---------------------------------------------------------
# PUNTO 4: DISPERSIÓN CON PROYECCIÓN DE HISTOGRAMAS EN CADA EJE
# ---------------------------------------------------------

fig = plt.figure(figsize=(8, 8))
gs = gridspec.GridSpec(4, 4)

# Definir paneles: centro (dispersión), arriba (hist X), derecha (hist Y)
ax_main = fig.add_subplot(gs[1:4, 0:3])
ax_xhist = fig.add_subplot(gs[0, 0:3], sharex=ax_main)
ax_yhist = fig.add_subplot(gs[1:4, 3], sharey=ax_main)

# 1. Dispersión principal
ax_main.scatter(X, Y, alpha=0.3, color='#1f77b4', s=15)
ax_main.set_xlabel('Perigeo $X$ (km)', fontsize=11)
ax_main.set_ylabel('Apogeo $Y$ (km)', fontsize=11)
ax_main.grid(True, linestyle='--', alpha=0.5)

# 2. Histograma de proyección en el eje X (arriba)
ax_xhist.hist(X, bins=30, color='#6baed6', edgecolor='black', alpha=0.7)
ax_xhist.tick_params(labelbottom=False)
ax_xhist.set_ylabel('Frecuencia $X$')
ax_xhist.grid(True, linestyle='--', alpha=0.5)

# 3. Histograma de proyección en el eje Y (derecha)
ax_yhist.hist(Y, bins=30, orientation='horizontal', color='#6baed6', edgecolor='black', alpha=0.7)
ax_yhist.tick_params(labelleft=False)
ax_yhist.set_xlabel('Frecuencia $Y$')
ax_yhist.grid(True, linestyle='--', alpha=0.5)

plt.suptitle('Punto 4: Proyección Marginal de $X$ e $Y$', fontsize=13, fontweight='bold', y=0.93)
plt.show()
```

```python
# Ajuste lineal por mínimos cuadrados con covarianza
p, cov = np.polyfit(X, Y, 1, cov=True)
m, b = p[0], p[1]
delta_m = np.sqrt(cov[0, 0])
delta_b = np.sqrt(cov[1, 1])
cov_mb = cov[0, 1]

# Cálculo de Chi-cuadrada (bondad de ajuste)
Y_modelo = m * X + b
residuals = Y - Y_modelo
sigma_y_i = np.std(residuals, ddof=2)
chi2 = np.sum(((Y - Y_modelo) / sigma_y_i) ** 2)
df = len(X) - 2
chi2_reducido = chi2 / df

print("=== PARÁMETROS Y CHI-CUADRADA ===")
print(
    f"Ecuación de ajuste: y = ({m:.4f} ± {delta_m:.4f})x + ({b:.2f} ± {delta_b:.2f})"
)
print(f"Estadístico Chi²: {chi2:.4f}")
print(f"Grados de libertad (df): {df}")
print(f"Chi² Reducido: {chi2_reducido:.4f}")
```

```python
# Definir un rango de valores de x para evaluar la propagación en toda la recta
x_line = np.linspace(min(X), max(X), 100)
y_line = m * x_line + b

# Propagación de error considerando la incertidumbre del ajuste
# MÁS la dispersión natural de los datos (sigma_y_i) para que la banda sea visible
delta_y_line = np.sqrt(
    (x_line**2) * (delta_m**2) + (delta_b**2) + 2 * x_line * cov_mb
    + sigma_y_i**2
)

# Gráfica del ajuste con su banda de error visible
plt.figure(figsize=(7, 4.5))
plt.scatter(
    X,
    Y,
    alpha=0.25,
    color="gray",
    s=15,
    label="Datos experimentales (Satélites LEO)",
)
plt.plot(
    x_line,
    y_line,
    color="darkblue",
    lw=2,
    label=rf"Ajuste lineal ($\chi^2_{{red}} \approx {chi2_reducido:.2f}$)",
)

# Banda sombreada bien visible
plt.fill_between(
    x_line,
    y_line - delta_y_line,
    y_line + delta_y_line,
    color="deepskyblue",
    alpha=0.35,  # Un poco más opaca para que resalte perfecto
    label=r"Banda de confianza / predicción",
)

plt.title(
    "Ajuste Lineal con Banda de Confianza Visible",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Perigeo X (km)")
plt.ylabel("Apogeo Y (km)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
